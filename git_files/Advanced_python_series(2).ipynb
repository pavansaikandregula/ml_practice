# public,private and protected:

```python
#all the class variables are public
class Car:
    def __init__(self, windows, doors , enginetype):
        self.windows = windows
        self . doors = doors
        self.enginetype = enginetype
        
```

```python
audi = Car(4,5,"diesel")
dir(audi)
```

```python
#all the class variables are protected
class Car2:
    def __init__(self, windows, doors , enginetype):
        self._windows = windows
        self._doors = doors
        self._enginetype = enginetype
#can be accessed by sub classes
```

```python
audi3 = Car2(6,7,9)
dir(audi3)
```

```python
#the python can't force it.
audi3._windows
```

```python
class Truck(Car2):
    def __init__(self,windows,doors,enginetype,horsepower):
        super(). __init__(windows,doors,enginetype) #
        self.horsepower = horsepower
```

```python
truck = Truck(4,5,6,7)
dir(truck)
```

```python
#private class
class Laptop:
    def __init__(self,username,password):
        self.username = username
        self.__password = password
```

```python
laptop = Laptop("sampath",1234)
```

```python
dir(laptop)
```

2. Protected Members

Convention: start with a single underscore.

class Car:
    def __init__(self):
        self._windows = 4

Access:

c = Car()
print(c._windows)

Output:

4

Still accessible!

Meaning

The underscore tells programmers:

"Please don't use this directly outside the class.
It's intended for the class and its subclasses."

But Python does not stop you.

Private Members

Convention: start with double underscore.

class Car:
    def __init__(self):
        self.__windows = 4

Access:

c = Car()
print(c.__windows)

Output:

AttributeError

Why?

Python changes the name internally.

__windows
    ↓
_Car__windows

This is called name mangling.

Proof
class Car:
    def __init__(self):
        self.__windows = 4

c = Car()

print(c._Car__windows)

Output:

4

So even private members are not truly hidden.

```python
| Access Type | Syntax     | Can Access Outside? | Purpose                  |
| ----------- | ---------- | ------------------- | ------------------------ |
| Public      | `self.x`   | ✅ Yes               | Normal use               |
| Protected   | `self._x`  | ✅ Yes               | Convention: internal use |
| Private     | `self.__x` | ⚠️ Not directly     | Name mangling            |
```
