```python
import matplotlib.pyplot as plt
%matplotlib inline
#no need to call show() everytime
```

```python
import numpy as np
```

```python
x = np.arange(0,10)
```

```python
y = np.arange(0,10)
```

marker represents the symbol,
line style represents the shape of line btw the points,
x label used for naming the x axis

```python
#scatter
y = x*x
plt.scatter(x,y,c='r',marker='v')
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.title("scatter plot")
plt.savefig('t_8_1.png')
```

```python
#plot
plt.plot(x,y,'r+--')#r indicates red,+ indicates marker,-- indicates how to connect
```

```python
#plot
plt.plot(x,y,'r',linestyle ='dashed',linewidth = 2,marker = 'o',markersize = 10)#r indicates red,+ indicates marker,-- indicates how to connect
plt.xlabel("x")
plt.ylabel("y")
plt.title("x and y")
```

**subplots:
plt.subplot(rows, columns, position) # syntax

```python
plt.subplot(2,2,1)
plt.plot(x,y,'ro--')
plt.xlabel("x-axis")
plt.subplot(2,2,2)
plt.plot(x,y,'bo--')
plt.subplot(2,2,3)
plt.plot(x,y,'go--')
plt.subplot(2,2,4)
plt.plot(x,y,'yo--')
```

```python
# matplotlib demo:
x = np.arange(0,11)
y = x * 3 +2
plt.plot(x,y,'r*--',linewidth = 5)
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.title("an demo")
```

```python
#sine curve:
x = np.arange(0,4*np.pi,0.1)
y = np.sin(x)
plt.plot(x,y,'ro--')
plt.title("sinx form")
```

```python
#cosine curve:
x = np.arange(0,4*np.pi,0.1)
y = np.cos(x)
plt.plot(x,y,'ro--')
plt.title("cos form")
```

```python
#subplot
plt.subplot(2,1,1)
x = np.arange(0,4*np.pi,0.1)
y = np.sin(x)
plt.plot(x,y,'ro--')
plt.title("sinx form")
plt.subplot(2,1,2)
x = np.arange(0,4*np.pi,0.1)
y = np.cos(x)
plt.plot(x,y,'g+--')
plt.title("cos form")
```

```python
#bar graph
#to understand for center and edge,check in o
x = [1,2,3]
y = [4,5,6]

x1 = [4,5,6]
y1 = [7,8,9]
plt.bar(x,y,align='center',color='g')
plt.bar(x1,y1,align='edge',color='r')
```

```python
labels = 'python','c++','ruby','java'
sizes = [215,130,245,210]
colors = ['gold','yellowgreen','blue','red']
explode = (0.4,0,0,0)

plt.pie(sizes,explode= explode,labels=labels,colors =colors,autopct =' %1.1f%%',shadow = True)
plt.axis('equal')
```
