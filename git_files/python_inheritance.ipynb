```python
###all class variables are public:
class Car():
    def __init__(self,windows,doors,enginetype):
        self.windows = windows
        self.doors = doors
        self.enginetype = enginetype
    def drive(self):
        print("the person drives the car")
```

```python
car = Car(4,5,"diesel")
```

```python
car.windows = 5
```

```python
class Truck(Car):
    def __init__(self,windows,doors,enginetype,space):
        super().__init__(windows,doors,enginetype)
        self.space = space
```

```python
truck = Truck(2,3,"petrol",10)
```
