```python
lst1= []
def lst_square(lst):
    for i in lst:
        lst1.append(i*i)
    return lst1    
```

```python
lst_square([1,2,3,4,5,6])
```

```python
#order matters
def get_info(name,age):
    return "hello {},your age is {}".format(name,age)
    
```

```python
#order matters
def get_info1(name,age):
    return "hello {},your age is {}".format(age,name)
```

```python
get_info1("sai",20)# fails here
```

```python
#order doesn't matters
def get_info2(name,age):
    return "hello {name1},your age is {age1}".format(age1 = age,name1 = name)
```
