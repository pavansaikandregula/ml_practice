```python
#let's try to except the error for adding different variable types:
try:
    a = 10
    b =" s "
    c = a + b
except NameError:
    print("the user has not defined the variable")
except TypeError:
    print("try to make the data type similar")
except Exception as e:
    print(e)
# the exception check from order      
```

```python
try:
    a = int(input("enter the number: "))
    b = int(input("enter the number: "))
    c = a + b
    d = a - b 
    e = a / b
    print(c," ",d," ",e,"  ")
except Exception as e:
    print(e)
print("hello")
```

```python
#try else
try:
    a = int(input("enter the number: "))
    b = int(input("enter the number: "))
    c = a + b
    d = a - b 
    e = a / b
except Exception as e:
    print(e)
else:
    print(c," ",d," ",e,"  ")
```

```python
#try else and finally.
#finally-> whatever happens, it will be printed
#try else
try:
    a = int(input("enter the number: "))
    b = int(input("enter the number: "))
    c = a + b
    d = a - b 
    e = a / b
except Exception as e:
    print(e)
else:
    print(c," ",d," ",e,"  ")
finally:
    print("execution is done")
```
