**In statistics, correlation is a measure that describes the strength and direction of the relationship between two variables...........**
**A correlation heatmap is a graphical visualization of a correlation matrix, using colored cells to represent the strength and direction of linear relationships between multiple variables. It is an essential exploratory data analysis (EDA) tool used to identify hidden patterns, validate assumptions, and prevent issues like multicollinearity in predictive modeling..............**
**accepts only with in int and float**

if the co relation is +ve, increase in one makes to increase in other..
negative means opposite

```python
df.corr(numeric_only=True)
```

```python
sns.heatmap(df.corr(numeric_only=True))
```

**A jointplot is a powerful data visualization function in the Seaborn Python library that combines a bivariate (two-variable) plot in the center with univariate (single-variable) marginal plots on the top and right margins**

```python
sns.jointplot(x = 'tip',y= 'total_bill',data = df,kind ='hex')
```

jooinplot() in Seaborn is used to show the relationship between two variables.

It combines:

Scatter plot (middle)
Distribution plot / histogram for each variable (sides)

So it gives:

relationship
trend
distribution
outliers

all together.

```python
sns.jointplot(x='tip',y='total_bill',data = df,kind = 'reg')
#draws probability density function..
```
