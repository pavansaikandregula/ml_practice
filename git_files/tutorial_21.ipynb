```python
#__init__ -> it's acts like an constructor
"""constructor: A constructor is a special member function of a class that is automatically called when an object is created.

Its main purpose is to initialize the object's data members"""
```

```python
#default constructor
class Car1:
    def __init__(self):
        print("i am default constructor")
```

```python
c1 =  Car1()
```

```python
#parameterized constructor:
class Car2:
    def __init__(self,windows,doors,seats):
        self.windows = windows
        self.door = doors
        self.seats = seats

    def display(self):
        print(self.door)
        print(self.windows)
        print(self.seats)

    def seat_type(self):
        if self.seats%2 == 0:
            return "the number of seats are {}, therefore it is even_type_seated".format(self.seats)
        else:
            return "the number of seats are {}, therefore it is odd_type_seated".format(self.seats)
```

```python
c2 = Car2(10,20,30)
```

```python
c2.display()
```

```python
c2.seat_type()
```

# how self works:

```python
class Student:
    def __init__(self, name):
        self.name = name

    def display(self):
        print(self.name)

s1 = Student("Pavan")
s1.display()
```

in the above code: 
When you write:

s1.display()

Python internally does:

Student.display(s1)

So:

self = s1

inside the display() method. 

Why use self?

To access the object's own variables and methods.

class Student:
    def __init__(self, name):
        self.name = name

    def show(self):
        print(self.name)

Here:

self.name

means:

"the name belonging to this particular object."

Flow
Student("Pavan")
      |
      v
name = "Pavan"   (parameter)
      |
      v
self.name = name
      |
      v
object attribute name = "Pavan"
