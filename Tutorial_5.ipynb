.loc[] can access: ROWS,COLUMNS, BOTH ROWS AND COLUMNS \n
// TO ACCESS COLUMNS:\n
df.loc[0:1,"name"]
for cols -> []
for rows -- > iloc [same like indexing]

| Code                        | Output Type |
| --------------------------- | ----------- |
| `d1.loc['ROW1']`            | Series      |
| `d1.loc[['ROW1']]`          | DataFrame   |
| `d1['col1']`                | Series      |
| `d1[['col1']]`              | DataFrame   |
| `d1.loc['ROW1','col1']`     | Scalar      |
| `d1.loc[['ROW1'],['col1']]` | DataFrame   |
Rule:
Selecting a single row or single column → returns a Series
Selecting multiple rows or multiple columns → returns a DataFrame
